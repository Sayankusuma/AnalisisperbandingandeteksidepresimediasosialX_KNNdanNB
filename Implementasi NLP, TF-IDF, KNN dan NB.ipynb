```python
import pandas as pd
# Membaca dataset
df = pd.read_csv("DatasetLabel.csv")

# Melihat jumlah data
print("Jumlah data:", len(df))

# Melihat nama kolom
print("\nNama kolom:")
print(df.columns.tolist())

# Melihat 5 data pertama
display(df.head())
```

```python
import pandas as pd
import ast
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

CLEANING

```python
import string
def cleaning(text):
    text = str(text)
    # Menghapus URL
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    # Menghapus mention @username
    text = re.sub(r'@\w+', '', text)
    # Menghapus tanda #
    # Kata setelah # tetap dipertahankan
    text = re.sub(r'#', '', text)
    # Menghapus angka
    text = re.sub(r'\d+', '', text)
    # Menghapus tanda baca
    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )
    # Menghapus karakter selain huruf dan spasi
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    # Menghapus spasi berlebihan
    text = re.sub(r'\s+', ' ', text).strip()
    return text
df['cleaning'] = df['full_text'].apply(cleaning)
```

CASE FOLDING

```python
def case_folding(text):
    return text.lower()

df['case_folding'] = df['cleaning'].apply(case_folding)
df[['cleaning', 'case_folding']].head(10)
```

# TOKENEZING

```python
def tokenizing(text):
    return text.split()

df['tokenizing'] = df['case_folding'].apply(tokenizing)
df[['case_folding', 'tokenizing']].head(10)
```

STOPWARD REMOVAL

```python
factory = StopWordRemoverFactory()

stopwords = set(factory.get_stop_words())

# Lindungi kata negasi agar tidak terhapus
kata_penting = {
    'tidak', 'bukan', 'belum', 'jangan', 'tak', 'di','ke', 'ter','me', 'meng', 'ber',
}

stopwords = stopwords - kata_penting

def stopword_removal(tokens):
    return [
        kata for kata in tokens
        if kata not in stopwords
    ]

df['stopword_removal'] = df['tokenizing'].apply(
    stopword_removal
)

df[['tokenizing', 'stopword_removal']].head(10)
```

STEMMING

```python
factory_stemmer = StemmerFactory()
stemmer = factory_stemmer.create_stemmer()
def stemming(tokens):
    return [
        stemmer.stem(word)
        for word in tokens
    ]
```

```python
df['stemming'] = df['stopword_removal'].apply(stemming)
```

```python
df['text_preprocessed'] = df['stemming'].apply(
    lambda x: ' '.join(x)
)
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Menentukan nilai K yang akan diuji
nilai_k = [3, 5, 7]

# Menyimpan model dan hasil prediksi
model_knn = {}
prediksi_knn = {}

for k in nilai_k:

    print(f"\n===== KNN dengan K = {k} =====")

    # Membuat model KNN
    knn = KNeighborsClassifier(
        n_neighbors=k,
        metric='euclidean'
    )

    # Training model
    knn.fit(X_train_tfidf, y_train)

    # Prediksi data testing
    y_pred = knn.predict(X_test_tfidf)

    # Menyimpan model
    model_knn[k] = knn

    # Menyimpan hasil prediksi
    prediksi_knn[k] = y_pred

    print("Model berhasil dilatih.")
    print("Jumlah data yang diprediksi:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

hasil_knn = []

for k in nilai_k:

    # Mengambil hasil prediksi K
    y_pred = prediksi_knn[k]

    # Menghitung metrik
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    # Menyimpan hasil
    hasil_knn.append({
        "Metode": "KNN",
        "K": k,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })
```

```python
#BINARY
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
# DAFTAR NILAI ALPHA
daftar_alpha = [
   0.001, 0.002, 0.005,
    0.01, 0.02, 0.05,
    0.1, 0.2,
]
hasil_alpha = []
#  PENGUJIAN SETIAP NILAI ALPHA
for alpha in daftar_alpha:
    # Membuat model Naive Bayes
    nb = MultinomialNB(alpha=alpha)
    # Training menggunakan data training yang sudah ada
    nb.fit(X_train_tfidf, y_train)
    # Prediksi menggunakan data testing yang sudah ada
    y_pred = nb.predict(X_test_tfidf)
    # Menghitung evaluasi
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    # Menyimpan hasil
    hasil_alpha.append({
        "Alpha": alpha,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

#  MEMBUAT TABEL HASIL
tabel_alpha = pd.DataFrame(hasil_alpha)

print("===== HASIL OPTIMASI PARAMETER NAIVE BAYES =====")

display(
    tabel_alpha.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}"
    })
)

# MENAMPILKAN ALPHA DENGAN AKURASI TERTINGGI

index_terbaik = tabel_alpha["Accuracy"].idxmax()

print("\n===== PARAMETER TERBAIK =====")
print(tabel_alpha.loc[index_terbaik])
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    roc_curve,
    roc_auc_score
)

# ==========================================
# 1. MELATIH MODEL KNN (K = 2369)
# ==========================================

knn_best = KNeighborsClassifier(n_neighbors=1761)

knn_best.fit(X_train_tfidf, y_train)

# Mengambil probabilitas kelas positif (depresi)
y_prob_knn = knn_best.predict_proba(X_test_tfidf)[:, 1]


# ==========================================
# 2. MELATIH MODEL NAIVE BAYES
# ==========================================

nb_best = MultinomialNB()

nb_best.fit(X_train_tfidf, y_train)

# Mengambil probabilitas kelas positif (depresi)
y_prob_nb = nb_best.predict_proba(X_test_tfidf)[:, 1]


# ==========================================
# 3. MENGHITUNG NILAI AUC
# ==========================================

auc_knn = roc_auc_score(y_test, y_prob_knn)

auc_nb = roc_auc_score(y_test, y_prob_nb)

# Menghitung selisih AUC
selisih_auc = abs(auc_knn - auc_nb)


# ==========================================
# 4. MENAMPILKAN TABEL AUC
# ==========================================

tabel_auc = pd.DataFrame({
    "Algoritma": [
        "K-Nearest Neighbors (KNN)",
        "Multinomial Naive Bayes"
    ],
    "Parameter": [
        "K = 2369",
        "Default"
    ],
    "AUC": [
        auc_knn,
        auc_nb
    ]
})

print("HASIL PERBANDINGAN NILAI AUC")

display(
    tabel_auc.style.format({
        "AUC": "{:.4f}"
    }).hide(axis="index")
)

print("\nSelisih Nilai AUC")
print(f"Selisih AUC : {selisih_auc:.4f}")
print(f"Selisih (%) : {selisih_auc * 100:.2f} poin persentase")


# ==========================================
# 5. MENGHITUNG KURVA ROC
# ==========================================

fpr_knn, tpr_knn, threshold_knn = roc_curve(
    y_test, y_prob_knn, pos_label=1
)

fpr_nb, tpr_nb, threshold_nb = roc_curve(
    y_test, y_prob_nb, pos_label=1
)


# ==========================================
# 6. VISUALISASI KURVA ROC
# ==========================================

plt.figure(figsize=(9, 7))

# Kurva ROC KNN
plt.plot(
    fpr_knn,
    tpr_knn,
    label=f"KNN (K=2369) - AUC = {auc_knn:.4f}",
    linewidth=2
)

# Kurva ROC Naive Bayes
plt.plot(
    fpr_nb,
    tpr_nb,
    label=f"Naive Bayes - AUC = {auc_nb:.4f}",
    linewidth=2
)

# Garis diagonal (Random Classifier)
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("Perbandingan Kurva ROC KNN dan Naive Bayes")

plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.xlim([0, 1])
plt.ylim([0, 1.05])

plt.tight_layout()
plt.show()


# ==========================================
# 7. KESIMPULAN MODEL BERDASARKAN AUC
# ==========================================

print("\nKESIMPULAN")

if auc_knn > auc_nb:
    print("KNN memiliki nilai AUC lebih tinggi.")
    print(f"KNN unggul sebesar {selisih_auc:.4f} dibandingkan Naive Bayes.")
elif auc_nb > auc_knn:
    print("Naive Bayes memiliki nilai AUC lebih tinggi.")
    print(f"Naive Bayes unggul sebesar {selisih_auc:.4f} dibandingkan KNN.")
else:
    print("Kedua model memiliki nilai AUC yang sama.")
```
